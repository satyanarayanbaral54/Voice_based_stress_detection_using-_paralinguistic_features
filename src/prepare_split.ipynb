```python
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder, StandardScaler
import pickle

X_seq = np.load("../data/X_seq.npy")
X_agg = np.load("../data/X_agg.npy")
y = np.load("../data/y.npy", allow_pickle=True)

le = LabelEncoder()
y_encoded = le.fit_transform(y)

scaler = StandardScaler()
X_agg_scaled = scaler.fit_transform(X_agg)

X_seq_train, X_seq_test, X_agg_train, X_agg_test, y_train, y_test = train_test_split(
    X_seq, X_agg_scaled, y_encoded, test_size=0.2, random_state=42, stratify=y_encoded
)

np.save("../data/X_seq_train.npy", X_seq_train)
np.save("../data/X_seq_test.npy", X_seq_test)
np.save("../data/X_agg_train.npy", X_agg_train)
np.save("../data/X_agg_test.npy", X_agg_test)
np.save("../data/y_train.npy", y_train)
np.save("../data/y_test.npy", y_test)

with open("../models/label_encoder.pkl", "wb") as f:
    pickle.dump(le, f)
with open("../models/scaler.pkl", "wb") as f:
    pickle.dump(scaler, f)

print("Classes:", le.classes_)
print("Train size:", len(y_train), "Test size:", len(y_test))
```
