```python
import numpy as np

# Load the same train/test data you already created
X_seq_train = np.load("../data/X_seq_train.npy")
X_seq_test = np.load("../data/X_seq_test.npy")

# Current shape: (samples, time_steps, mfcc_coefficients) e.g. (460, 130, 13)
# CNN needs: (samples, time_steps, mfcc_coefficients, channels) -> add 1 channel

X_cnn_train = X_seq_train[..., np.newaxis]
X_cnn_test = X_seq_test[..., np.newaxis]

np.save("../data/X_cnn_train.npy", X_cnn_train)
np.save("../data/X_cnn_test.npy", X_cnn_test)

print("CNN Train shape:", X_cnn_train.shape)
print("CNN Test shape:", X_cnn_test.shape)
```
